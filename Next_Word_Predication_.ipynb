```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
tokens=word_tokenize(document.lower())
tokens
```

```python
vocab={'<UNK>':0}
for token in Counter(tokens).keys():
  if token not in vocab:
    vocab[token]=len(vocab)
```

```python
vocab
```

```python
len(vocab)
```

```python
input_sentence=document.split('\n')
input_sentence
```

```python
def text_to_indicis(sentence,vocab):
  numerical_sentecne=[]
  for token in sentence:
    if token in vocab:
      numerical_sentecne.append(vocab[token])
    else:
      numerical_sentecne.append(vocab['<UNK>'])
  return numerical_sentecne
```

```python
input_numerical_sentence=[]
for sentence in input_sentence:
  input_numerical_sentence.append(text_to_indicis(word_tokenize(sentence.lower()),vocab))
```

```python
input_numerical_sentence
```

```python
training_sequnces=[]
for sentence in input_numerical_sentence:
  for i in range(1,len(sentence)):
    training_sequnces.append(sentence[:i+1])
```

```python
training_sequnces
```

```python
len_list=[]
for sequnce in training_sequnces:
  len_list.append(len(sequnce))
```

```python
max(len_list)
```

```python
padding_training_sequnce=[]
for sentence in training_sequnces:
  padding_training_sequnce.append([0]*(max(len_list)-len(sentence))+sentence)
```

```python
padding_training_sequnce[1]
```

```python
padding_training_sequnce=torch.tensor(padding_training_sequnce,dtype=torch.long)
```

```python
padding_training_sequnce
```

```python
X=padding_training_sequnce[:,:-1]
y=padding_training_sequnce[:,-1]
```

```python
len(X)
```

```python
len(y)
```

```python
class CustomDataset(Dataset):
  def __init__(self,X,y):
    self.X=X
    self.y=y
  def __len__(self):
    return self.X.shape[0]
  def __getitem__(self,idx):
    return self.X[idx],self.y[idx]
```

```python
dataset=CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataloader=DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
for input,output in dataloader:
  print(len(input))
```

```python

# vocab_size = len(vocab)
# embedding_dim = 100
# hidden_size = 128

class LSTMLanguageModel(nn.Module):
    def __init__(self, vocab_size=len(vocab), embedding_dim=100, hidden_size=128):
        super().__init__()

        self.embedding = nn.Embedding(vocab_size, embedding_dim)

        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_size,
            batch_first=True
        )

        self.fc = nn.Linear(hidden_size, vocab_size)

    def forward(self, x):
        x = self.embedding(x)
        x, _ = self.lstm(x)
        x = self.fc(x)
        return x

model = LSTMLanguageModel(
    vocab_size,
    embedding_dim,
    hidden_size
)

print(model)
```

```python
criterion=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(),lr=0.001)
```

```python
epochs = 50

for epoch in range(epochs):
    model.train()
    total_loss = 0

    for input_seq, target_seq in dataloader:

        optimizer.zero_grad()

        output = model(input_seq)

        output = output[:, -1, :]

        loss = criterion(output, target_seq)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(dataloader)

    print(f"Epoch {epoch + 1}, Loss: {avg_loss:.4f}")
```

```python
id_to_word = {idx: word for word, idx in vocab.items()}
```

```python
def predict_next_word(text):
    model.eval()

    words = text.lower().split()
    token_ids = []

    for word in words:
        if word in vocab:
            token_ids.append(vocab[word])
        else:
            token_ids.append(vocab.get("<UNK>", 0))

    input_tensor = torch.tensor( [token_ids],dtype=torch.long)

    with torch.no_grad():

        output = model(input_tensor)

        output = output[:, -1, :]

        predicted_id = torch.argmax(output, dim=1).item()

    return id_to_word.get(predicted_id, "<UNK>")
```

```python
text = "Strong waves "

prediction = predict_next_word(text)

print("Input:", text)
print("Predicted next word:", prediction)
```
